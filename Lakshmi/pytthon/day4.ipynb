```python
class Person:
    def __init__(self, name, age, blood_group, nationality):
        self.name = name
        self.age = age
        self.blood_group = blood_group
        self.nationality = nationality


# Child class
class Student(Person):
    def __init__(self, name, age, blood_group, nationality,
                 section, roll_number, percentage):

        # Calling parent class constructor
        super().__init__(name, age, blood_group, nationality)

        self.section = section
        self.roll_number = roll_number
        self.percentage = percentage

    def display(self):
        print("Student Name:", self.name)
        print("Age:", self.age)
        print("Blood Group:", self.blood_group)
        print("Nationality:", self.nationality)
        print("Section:", self.section)
        print("Roll Number:", self.roll_number)
        print("Percentage:", self.percentage)


# Creating Student object
student1 = Student(
    "Lakshmi",
    21,
    "O+",
    "Indian",
    "A",
    "49",
    85.5
)

student1.display()
```

```python
class Person:

    def __init__(self, name, father_name, usn, roll_number,
                 section, department, college_name,
                 tenth_result, twelfth_result, degree_percentage):

        # Private variables
        self.__name = name
        self.__father_name = father_name
        self.__usn = usn
        self.__roll_number = roll_number
        self.__section = section
        self.__department = department
        self.__college_name = college_name
        self.__tenth_result = tenth_result
        self.__twelfth_result = twelfth_result
        self.__degree_percentage = degree_percentage

    # Getter methods
    def get_name(self):
        return self.__name

    def get_father_name(self):
        return self.__father_name

    def get_usn(self):
        return self.__usn

    def get_roll_number(self):
        return self.__roll_number

    def get_section(self):
        return self.__section

    def get_department(self):
        return self.__department

    def get_college_name(self):
        return self.__college_name

    def get_tenth_result(self):
        return self.__tenth_result

    def get_twelfth_result(self):
        return self.__twelfth_result

    def get_degree_percentage(self):
        return self.__degree_percentage


# Creating object
p = Person(
    "Lakshmi Swamy",
    " Somanath Swamy",
    "3GN23EC049",
    "3527/23",
    "A",
    "ECE",
    "GNDEC Bidar",
    94.24,
    79.5,
    75.5
)

# Displaying details
print("Name:", p.get_name())
print("Father Name:", p.get_father_name())
print("USN:", p.get_usn())
print("Roll Number:", p.get_roll_number())
print("Section:", p.get_section())
print("Department:", p.get_department())
print("College Name:", p.get_college_name())
print("10th Result:", p.get_tenth_result(), "%")
print("12th Result:", p.get_twelfth_result(), "%")
print("Degree Percentage:", p.get_degree_percentage(), "%")
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Dog says: Bow Bow")

class Cat(Animal):
    def sound(self):
        print("Cat says: Meow")

# Create objects
d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Dog says: Bow Bow")


class Cat(Animal):

    def sound(self):
        print("Cat says: Meow")


# Creating objects
d = Dog()
c = Cat()

d.sound()
c.sound()
```

```python
arr = [50, 20, 40, 10, 30]

n = len(arr)

for i in range(n):
    for j in range(0, n - i - 1):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]

print("Sorted array:", arr)
```
